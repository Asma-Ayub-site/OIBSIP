```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

sns.set_theme(style="whitegrid")
```

```python
# ==========================================
#  LOAD DATASET & INITIAL INSPECTION
# ==========================================

# Load dataset
df = pd.read_csv("dataset/retail_sales_dataset.csv")

# ------------------------------------------
# Dataset Overview
# ------------------------------------------

overview = pd.DataFrame({
    'Metric': [
        'Number of Rows',
        'Number of Columns',
        'Missing Values',
        'Duplicate Rows'
    ],
    'Value': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

print("DATASET OVERVIEW")

display(
    overview.style
    .set_caption("Retail Sales Dataset Overview")
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# ------------------------------------------
# Column Information
# ------------------------------------------

column_info = pd.DataFrame({
    'Column Name': df.columns,
    'Data Type': df.dtypes.astype(str).values,
    'Non-Null Values': df.notnull().sum().values
})

print("\nCOLUMN INFORMATION")

display(
    column_info.style
    .set_caption("Dataset Column Information")
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# ------------------------------------------
# Missing Values
# ------------------------------------------

missing_values = pd.DataFrame({
    'Column': df.columns,
    'Missing Values': df.isnull().sum().values
})

print("\nMISSING VALUE CHECK")

display(
    missing_values.style
    .set_caption("Missing Values by Column")
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# ------------------------------------------
# First 5 Rows
# ------------------------------------------

print("\nSAMPLE DATA")

display(
    df.head().style
    .set_caption("First Five Records")
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)
```

```python
# ==========================================
#  DESCRIPTIVE STATISTICS
# ==========================================

numeric_columns = [
    'Age',
    'Quantity',
    'Price per Unit',
    'Total Amount'
]

statistics = pd.DataFrame({
    'Mean': df[numeric_columns].mean(),
    'Median': df[numeric_columns].median(),
    'Mode': df[numeric_columns].mode().iloc[0],
    'Standard Deviation': df[numeric_columns].std()
})

statistics = statistics.round(2)

print("DESCRIPTIVE STATISTICS")

display(
    statistics.style
    .set_caption("Descriptive Statistics of Numerical Variables")
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)
```

```python
# ==========================================
#  TIME SERIES ANALYSIS
# ==========================================

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Create Month and Quarter columns
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df['Quarter'] = df['Date'].dt.to_period('Q').astype(str)

# ------------------------------------------
# Monthly Sales
# ------------------------------------------

monthly_sales = (
    df.groupby('Month')['Total Amount']
    .sum()
    .sort_index()
)

print("MONTHLY SALES")

display(
    monthly_sales.to_frame('Total Sales')
    .style
    .set_caption("Monthly Sales Summary")
    .format("{:,.2f}")
    .set_properties(**{'text-align': 'center'})
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=16, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------
# Quarterly Sales
# ------------------------------------------

quarterly_sales = (
    df.groupby('Quarter')['Total Amount']
    .sum()
    .sort_index()
)

print("\nQUARTERLY SALES")

display(
    quarterly_sales.to_frame('Total Sales')
    .style
    .set_caption("Quarterly Sales Summary")
    .format("{:,.2f}")
    .set_properties(**{'text-align': 'center'})
)

plt.figure(figsize=(9, 5))

plt.plot(
    quarterly_sales.index,
    quarterly_sales.values,
    marker='o',
    linewidth=2
)

plt.title("Quarterly Sales Trend", fontsize=16, fontweight='bold')
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
### Monthly Sales — Observation

The monthly sales trend shows fluctuations throughout the observed period.
Sales vary considerably from month to month, indicating changes in purchasing activity over time.
The trend can help identify stronger and weaker sales periods for future planning and inventory management.
```

```python
### Quarterly Sales — Observation

The quarterly analysis provides a broader view of sales performance by grouping monthly transactions into quarters.
Differences between quarters indicate changes in overall purchasing activity and can help the business identify stronger
periods for planning marketing campaigns, inventory, and sales strategies.
```

```python
# ==========================================
#  CUSTOMER DEMOGRAPHICS
# ==========================================

# Create age groups
age_bins = [0, 20, 30, 40, 50, 60, 100]
age_labels = ['Under 20', '20-29', '30-39', '40-49', '50-59', '60+']

df['Age Group'] = pd.cut(
    df['Age'],
    bins=age_bins,
    labels=age_labels,
    right=False
)

# ------------------------------------------
# Age Group Distribution
# ------------------------------------------

age_distribution = (
    df['Age Group']
    .value_counts()
    .sort_index()
)

age_table = age_distribution.to_frame('Number of Customers')

print("AGE GROUP DISTRIBUTION")

display(
    age_table.style
    .set_caption("Customer Distribution by Age Group")
    .format("{:,.0f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# Age Group Chart
plt.figure(figsize=(10, 5))

age_distribution.plot(
    kind='bar',
    width=0.7
)

plt.title(
    "Customer Distribution by Age Group",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------
# Gender Distribution
# ------------------------------------------

gender_distribution = (
    df['Gender']
    .value_counts()
)

gender_table = gender_distribution.to_frame('Number of Customers')

print("\nGENDER DISTRIBUTION")

display(
    gender_table.style
    .set_caption("Customer Distribution by Gender")
    .format("{:,.0f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# Gender Chart
plt.figure(figsize=(8, 5))

gender_distribution.plot(
    kind='bar',
    width=0.6
)

plt.title(
    "Customer Distribution by Gender",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
### Age Group Analysis — Observation

The customer base is distributed across all age groups, with customers aged 40-49 representing the largest group at 223 customers,
followed closely by the 50-59 group with 221 customers.Customers under 20 represent the smallest group with 42 customers.
```

```python
### Gender Analysis — Observation

The gender distribution is relatively balanced. Female customers account for 510 records, while male customers account for 490 records. 
This indicates that the dataset does not show a substantial gender imbalance.
```

```python
# ==========================================
#  PRODUCT ANALYSIS
# ==========================================

# ------------------------------------------
# Quantity Sold by Product Category
# ------------------------------------------

category_quantity = (
    df.groupby('Product Category')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

print("QUANTITY SOLD BY PRODUCT CATEGORY")

display(
    category_quantity.to_frame('Total Quantity Sold')
    .style
    .set_caption("Product Category Performance by Quantity")
    .format("{:,.0f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# Quantity Chart
plt.figure(figsize=(9, 5))

category_quantity.plot(
    kind='bar',
    width=0.65
)

plt.title(
    "Quantity Sold by Product Category",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------
# Revenue by Product Category
# ------------------------------------------

category_revenue = (
    df.groupby('Product Category')['Total Amount']
    .sum()
    .sort_values(ascending=False)
)

print("\nREVENUE BY PRODUCT CATEGORY")

display(
    category_revenue.to_frame('Total Revenue')
    .style
    .set_caption("Revenue Performance by Product Category")
    .format("{:,.2f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# Revenue Chart
plt.figure(figsize=(9, 5))

category_revenue.plot(
    kind='bar',
    width=0.65
)

plt.title(
    "Revenue by Product Category",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
### Product Category — Quantity Observation

Clothing has the highest total quantity sold with 894 units, followed by Electronics with 849 units and Beauty with 771 units. 
This indicates that Clothing is the most frequently purchased category by sales volume.
```

```python
### Product Category — Revenue Observation

Electronics generated the highest revenue at 156,905, closely followed by Clothing at 155,580.
Beauty generated the lowest revenue at 143,515. 
Although Clothing had the highest quantity sold, Electronics generated slightly higher revenue.
```

```python
### Product Analysis — Dataset Limitation

The dataset does not contain individual product names. Therefore, a Top 10 individual product analysis could not be performed.
Product categories were analyzed instead using total quantity sold and revenue generated.
```

```python
# ==========================================
#  CORRELATION ANALYSIS
# ==========================================

# Select numerical variables
correlation_data = df[
    ['Age', 'Quantity', 'Price per Unit', 'Total Amount']
]

# Calculate correlation matrix
correlation_matrix = correlation_data.corr().round(2)

print("CORRELATION MATRIX")

display(
    correlation_matrix.style
    .set_caption("Correlation Matrix of Numerical Variables")
    .format("{:.2f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# ------------------------------------------
# Correlation Heatmap
# ------------------------------------------

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap of Numerical Variables",
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
### Correlation Analysis — Observation

The correlation analysis shows a strong positive relationship of 0.85 between Price per Unit and Total Amount.
Quantity has a moderate positive relationship with Total Amount at 0.37. 
Age has very weak relationships with the other numerical variables, with correlations close to zero.
This suggests that transaction value is more closely related to price and quantity than to customer age.
```

```python
# ==========================================
#  ADDITIONAL VISUALIZATION
# ==========================================

# Calculate average transaction amount by age group
age_group_spending = (
    df.groupby('Age Group', observed=True)['Total Amount']
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

# Display summary table
print("AVERAGE TRANSACTION AMOUNT BY AGE GROUP")

display(
    age_group_spending.to_frame('Average Transaction Amount')
    .style
    .set_caption("Average Transaction Amount by Age Group")
    .format("{:,.2f}")
    .set_properties(**{
        'text-align': 'center'
    })
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('font-weight', 'bold'),
                ('text-align', 'center')
            ]
        }
    ])
)

# ------------------------------------------
# Visualization
# ------------------------------------------

plt.figure(figsize=(10, 5))

age_group_spending.plot(
    kind='bar',
    width=0.65
)

plt.title(
    "Average Transaction Amount by Age Group",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Age Group")
plt.ylabel("Average Transaction Amount")

plt.xticks(rotation=0)

plt.grid(
    axis='y',
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
### Additional Visualization — Observation

The average transaction amount varies across age groups, showing that customer age may be associated with differences in spending behaviour.
The age group with the highest average transaction amount represents a potentially valuable customer segment for targeted marketing and promotional strategies.
However, the correlation analysis indicates that age has only a very weak linear relationship with total transaction amount, so age should be considered alongside other customer and transaction characteristics.
```

```python
#  CONCLUSION AND BUSINESS RECOMMENDATIONS
## Key Findings
1. Monthly and quarterly sales show noticeable fluctuations throughout the analysis period. Q4 2023 recorded the highest quarterly sales at 126,190, while Q3 2023 recorded the lowest sales among the complete quarters at 96,045.
2. The customer base is relatively balanced by gender, with 510 female and 490 male customers. The 40-49 age group has the highest representation with 223 customers.
3. Clothing recorded the highest sales quantity with 894 units, while Electronics generated the highest revenue at 156,905.
4. The correlation analysis shows a strong positive relationship between Price per Unit and Total Amount (0.85), while Quantity has a moderate positive relationship with Total Amount (0.37).
5. Age has very weak linear relationships with the main numerical sales variables, suggesting that age alone is not a strong predictor of transaction value.

## Business Recommendations

### 1. Strengthen Electronics Marketing
Electronics generated the highest revenue among the product categories. The business should consider targeted promotions, product bundles, and cross-selling strategies for Electronics to further increase revenue.
### 2. Maintain Strong Clothing Sales
Clothing recorded the highest quantity sold. The business should maintain adequate inventory levels and use seasonal promotions to capitalize on the strong demand for Clothing.
### 3. Use Customer Segmentation
Customer demographics and purchasing behaviour can be used to create targeted marketing campaigns. Different offers can be developed for high-value age groups and customer segments to encourage repeat purchases and increase average transaction value.

## Dataset Limitation
The dataset does not contain individual product names. Therefore, a Top 10 individual product analysis could not be performed. Product categories were analyzed instead using total quantity sold and revenue generated.

## Overall Conclusion
The exploratory data analysis provides useful insights into sales trends, customer demographics, product category performance, and relationships between numerical variables. These findings can support better inventory planning, targeted marketing, and revenue-focused business decisions.
```
